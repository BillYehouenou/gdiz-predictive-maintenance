# Limites

Ici, je recense tous les axes d'améliorations que je n'ai pas pu explorer :

1. **Données synthétiques, pas des capteurs GDIZ réels** : le générateur reproduit des dynamiques physiques plausibles (calendrier saisonnier béninois, réseau SBEE, cycles Harmattan) mais reste une simulation calibrée.

2. **OSF (surcharge mécanique) reste le point faible** : environ 60 % de recall, limité par la rareté des exemples (23-30 pannes OSF selon le split).

3. **Préavis médian de 12,5h, pas plusieurs jours** : suffisant pour planifier une intervention le jour même, insuffisant si une pièce devrait être commandée à l'avance.

4. **Pas d'authentification sur l'API** : Pas nécessaire pour le périmètre actuel, mais à prévoir si l'API devrait être exposée à des tiers.

5. **Cause probable issues de familles physiques (SHAP)** : l'étage de diagnostic multi-classe testé séparément était trop instable pour être exposé (accuracy validation 46,8 %, OSF quasi jamais reconnu) ; le dashboard affiche plutôt "quel phénomène physique pousse cette alerte" (usure, surchauffe, surcharge mécanique...), une question différente mais toujours fiable.
